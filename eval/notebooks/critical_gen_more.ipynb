```python
import torch
from diffusers import FluxPipeline
pipe_krea = FluxPipeline.from_pretrained("black-forest-labs/FLUX.1-Krea-dev", torch_dtype=torch.bfloat16)
pipe_krea = pipe_krea.to("cuda")

pipe_dance = FluxPipeline.from_pretrained("weathon/dance_flux", torch_dtype=torch.bfloat16)
pipe_dance = pipe_dance.to("cuda")
```

```python
prompts = [
    "An anti-war image. A mother holding a baby sitting in front of a completely distorted ruin during a war. Houses behind them are broken. The mother shows a facial expression of hopeless.",
    "A river with trash floating on it. The water has a dark color. There are factories next to the river with high chimneys releasing dark smoke and pipes releasing wastewater into the river.",
    "Someone holding a phone using their weak fingers, with their head surrounded by internet evil emojies, and their face showing a drained expression. The background is chaos Tsunami made from data, almost engulfing the subject. The sunlight has a clashing color.",
    "A college student holding a book in an old and broken classroom with a leaky roof, shows in black and white on the student's side. However, the book has no pages but only the frame. In front of them, there is a wide and deep trench with lava in it. The student shows a sad face looking at a fancy college campus on the other side of the trench, which is colorful."
]
```

```python
# torch.compile(pipe_krea.transformer,
# pipe_krea.transformer.to(memory_format=torch.channels_last)
```

```python
pipe_dance(
        prompt="A dark polluted industrial landscape where factories spew thick smoke into a blurry sky. The colors are harsh and clashing—acidic green, brown, and gray—creating a visually unpleasant and distorted scene. The main subject is unclear, blending into the chaotic surroundings. Everything feels broken and rough, evoking emptiness and decay.",
        num_inference_steps=28
    ).images[0]
```

```python
pipe_krea = pipe_krea.to("cuda")
```

```python
ds = []
for _ in range(5):
  for i in range(len(prompts)):
    critical_prompt = prompts[i]
    critical_image = pipe_krea(
        prompt=critical_prompt,
        num_inference_steps=28
    ).images[0]
    softer_image = pipe_dance(
        prompt=critical_prompt,
        num_inference_steps=28
    ).images[0]
    ds.append({
      "prompt": critical_prompt,
      "DanceFlux": softer_image,
      "Flux Krea": critical_image,
    })
```

```python
from datasets import Dataset
ds = Dataset.from_list(ds)
```

```python
import numpy as np
def hstack(images):
    if len(images) == 0:
        raise ValueError("Need 0 or more images")

    if isinstance(images[0], np.ndarray):
        images = [Image.fromarray(img) for img in images]
    width = sum([img.size[0] for img in images])
    height = max([img.size[1] for img in images])
    stacked = Image.new(images[0].mode, (width, height))

    x_pos = 0
    for img in images:
        stacked.paste(img, (x_pos, 0))
        x_pos += img.size[0]
    return stacked
```
